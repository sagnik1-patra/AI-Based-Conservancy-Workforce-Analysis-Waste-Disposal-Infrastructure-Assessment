```python
import json
import pickle
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore", category=FutureWarning)

sns.set_theme(style="whitegrid")

# ================================================================
# 1. PROJECT PATH CONFIGURATION
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Conservancy Workforce Analysis, Waste Disposal Infrastructure Assessment, and Resource Optimization System"
)

CSV_PATH = BASE_DIR / (
    "number_of conservancy_workers_engaged_in_conservancy_"
    "garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

def output_path(filename):
    return BASE_DIR / ("pso_" + filename)

# ================================================================
# 2. PSO HYPERPARAMETERS
# ================================================================

SEED = 42
rng = np.random.default_rng(SEED)

SWARM_SIZE = 40
MAX_ITERATIONS = 150

INERTIA_START = 0.9
INERTIA_END = 0.4

COGNITIVE_COEFFICIENT = 1.8
SOCIAL_COEFFICIENT = 1.8

VELOCITY_LIMIT = 3.0

BASE_YEAR = 2018
FORECAST_YEARS = 10

WORKFORCE_GROWTH_RATE = 0.04
SITE_GROWTH_RATE = 0.03

print("=" * 70)
print("PARTICLE SWARM OPTIMIZATION - CONSERVANCY WORKFORCE")
print("=" * 70)

# ================================================================
# 3. LOAD DATASET
# ================================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head().to_string(index=False))

print("\nDataset Shape:", df.shape)

# ================================================================
# 4. VALIDATE COLUMNS
# ================================================================

COLUMN_MAPPING = {
    "Number of workers engaged in conservancy work":
        "conservancy_workers",

    "Number of garbage and sewage disposal dump site":
        "disposal_sites",

    "Number of persons engaged in other connected works related with conservancy":
        "supporting_workers"
}

missing = [
    col for col in COLUMN_MAPPING
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

data = df.rename(columns=COLUMN_MAPPING).copy()

required_columns = [
    "conservancy_workers",
    "supporting_workers",
    "disposal_sites"
]

for col in required_columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

if data[required_columns].isna().any().any():
    raise ValueError(
        "Dataset contains invalid numeric values."
    )

if (data[required_columns] < 0).any().any():
    raise ValueError(
        "Worker and site counts cannot be negative."
    )

numeric_values = data[required_columns].to_numpy()

if not np.all(
    numeric_values == np.floor(numeric_values)
):
    raise ValueError(
        "Worker and site counts must be integers."
    )

data[required_columns] = (
    data[required_columns].astype(int)
)

# ================================================================
# 5. WORKFORCE STATISTICS
# ================================================================

total_conservancy = int(
    data["conservancy_workers"].sum()
)

total_supporting = int(
    data["supporting_workers"].sum()
)

total_sites = int(
    data["disposal_sites"].sum()
)

total_workers = (
    total_conservancy + total_supporting
)

if total_sites <= 0:
    raise ValueError(
        "At least one disposal site is required."
    )

print("\nWORKFORCE STATISTICS")
print("-" * 50)

print("Conservancy Workers:", total_conservancy)
print("Supporting Workers:", total_supporting)
print("Disposal Sites:", total_sites)
print("Total Workers:", total_workers)

print(
    "Average Workers Per Site:",
    round(total_workers / total_sites, 4)
)

# ================================================================
# 6. WORKLOAD ASSUMPTIONS
# ================================================================

# Site-specific workloads are not available.
# For a controlled experiment, assume weights
# increasing from 1.0 to 2.0.
#
# Use the same weights and objective as the AIS experiment
# for meaningful algorithm comparison.

workload_weights = np.linspace(
    1.0,
    2.0,
    total_sites
)

workload_weights = (
    workload_weights / workload_weights.sum()
)

target_conservancy = (
    total_conservancy * workload_weights
)

target_supporting = (
    total_supporting * workload_weights
)

# ================================================================
# 7. INTEGER ALLOCATION UTILITIES
# ================================================================

def equal_allocate(total, sites):

    base, remainder = divmod(
        int(total), int(sites)
    )

    result = np.full(
        sites,
        base,
        dtype=int
    )

    result[:remainder] += 1

    return result


def proportional_allocate(total, weights):

    weights = np.asarray(weights, dtype=float)
    weights = weights / weights.sum()

    raw = total * weights

    result = np.floor(raw).astype(int)

    remainder = int(
        total - result.sum()
    )

    fractions = raw - result

    order = np.argsort(-fractions)

    result[order[:remainder]] += 1

    return result


def random_allocate(total, sites, generator):

    if total == 0:
        return np.zeros(sites, dtype=int)

    probabilities = generator.dirichlet(
        np.ones(sites)
    )

    return generator.multinomial(
        total,
        probabilities
    )


# ================================================================
# 8. PARTICLE DECODING
# ================================================================

# Each PSO particle contains continuous positions.
#
# The positions are converted into nonnegative integer
# allocations while preserving workforce totals.
#
# Particle dimensions:
# First N positions: conservancy allocation scores
# Next N positions: supporting allocation scores

def decode_particle(
    particle,
    conservancy_count,
    supporting_count,
    sites
):

    first = particle[:sites]
    second = particle[sites:]

    # Softmax converts continuous scores to positive weights.
    # Subtract maximum for numerical stability.

    first_exp = np.exp(
        np.clip(first - np.max(first), -50, 0)
    )

    second_exp = np.exp(
        np.clip(second - np.max(second), -50, 0)
    )

    first_weights = (
        first_exp / first_exp.sum()
    )

    second_weights = (
        second_exp / second_exp.sum()
    )

    conservancy_allocation = proportional_allocate(
        conservancy_count,
        first_weights
    )

    supporting_allocation = proportional_allocate(
        supporting_count,
        second_weights
    )

    return np.vstack([
        conservancy_allocation,
        supporting_allocation
    ])


# ================================================================
# 9. FITNESS AND COST FUNCTIONS
# ================================================================

def allocation_cost(
    allocation,
    target_c,
    target_s
):
    """
    Same objective as the AIS implementation.

    Minimize mean absolute allocation deviation.
    """

    c = allocation[0]
    s = allocation[1]

    error_c = np.mean(
        np.abs(c - target_c)
    )

    error_s = np.mean(
        np.abs(s - target_s)
    )

    return float(error_c + error_s)


def fitness_from_cost(cost):

    return 1.0 / (1.0 + cost)


# ================================================================
# 10. PSO OPTIMIZATION FUNCTION
# ================================================================

def run_pso(
    conservancy_count,
    supporting_count,
    sites,
    weights,
    swarm_size=40,
    iterations=150,
    generator=None
):
    """
    Particle Swarm Optimization.

    Objective:
    Minimize deviation from hypothetical
    workload-weighted staffing targets.

    Returns:
    - Best integer allocation
    - Best cost
    - Best fitness
    - Convergence history
    """

    if generator is None:
        generator = np.random.default_rng(SEED)

    weights = np.asarray(weights, dtype=float)
    weights = weights / weights.sum()

    target_c = conservancy_count * weights
    target_s = supporting_count * weights

    dimensions = 2 * sites

    # ------------------------------------------------------------
    # Initialize particle positions
    # ------------------------------------------------------------

    positions = generator.normal(
        loc=0.0,
        scale=1.0,
        size=(swarm_size, dimensions)
    )

    velocities = generator.uniform(
        low=-0.5,
        high=0.5,
        size=(swarm_size, dimensions)
    )

    # Initialize one particle close to proportional allocation.
    # Logarithms are suitable because decoding uses softmax.

    initial_scores = np.log(
        np.maximum(weights, 1e-12)
    )

    positions[0] = np.concatenate([
        initial_scores,
        initial_scores
    ])

    # Equal allocation baseline
    positions[1] = np.zeros(dimensions)

    # ------------------------------------------------------------
    # Evaluate initial swarm
    # ------------------------------------------------------------

    personal_best_positions = positions.copy()

    personal_best_costs = np.full(
        swarm_size,
        np.inf
    )

    global_best_position = None
    global_best_allocation = None
    global_best_cost = float("inf")

    # Seed global memory with a feasible proportional baseline.
    baseline_allocation = np.vstack([
        proportional_allocate(
            conservancy_count, weights
        ),
        proportional_allocate(
            supporting_count, weights
        )
    ])

    global_best_allocation = baseline_allocation.copy()

    global_best_cost = allocation_cost(
        baseline_allocation,
        target_c,
        target_s
    )

    global_best_position = positions[0].copy()

    history = []

    # ------------------------------------------------------------
    # Main PSO loop
    # ------------------------------------------------------------

    for iteration in range(iterations):

        inertia = (
            INERTIA_START
            -
            (INERTIA_START - INERTIA_END)
            * iteration
            / max(iterations - 1, 1)
        )

        current_costs = []

        for i in range(swarm_size):

            allocation = decode_particle(
                positions[i],
                conservancy_count,
                supporting_count,
                sites
            )

            cost = allocation_cost(
                allocation,
                target_c,
                target_s
            )

            current_costs.append(cost)

            if cost < personal_best_costs[i]:

                personal_best_costs[i] = cost

                personal_best_positions[i] = (
                    positions[i].copy()
                )

            if cost < global_best_cost:

                global_best_cost = cost

                global_best_position = (
                    positions[i].copy()
                )

                global_best_allocation = (
                    allocation.copy()
                )

        # --------------------------------------------------------
        # Velocity and position update
        # --------------------------------------------------------

        r1 = generator.random(
            (swarm_size, dimensions)
        )

        r2 = generator.random(
            (swarm_size, dimensions)
        )

        cognitive = (
            COGNITIVE_COEFFICIENT
            * r1
            * (
                personal_best_positions
                - positions
            )
        )

        social = (
            SOCIAL_COEFFICIENT
            * r2
            * (
                global_best_position
                - positions
            )
        )

        velocities = (
            inertia * velocities
            + cognitive
            + social
        )

        velocities = np.clip(
            velocities,
            -VELOCITY_LIMIT,
            VELOCITY_LIMIT
        )

        positions = positions + velocities

        # Prevent extreme numerical scores.
        positions = np.clip(
            positions,
            -15,
            15
        )

        history.append({
            "iteration": iteration + 1,
            "best_cost": global_best_cost,
            "mean_cost": float(
                np.mean(current_costs)
            ),
            "best_fitness": fitness_from_cost(
                global_best_cost
            ),
            "inertia": inertia
        })

        if (
            iteration == 0
            or (iteration + 1) % 10 == 0
        ):

            print(
                f"Iteration {iteration + 1:3d} | "
                f"Best Cost: {global_best_cost:.6f} | "
                f"Fitness: "
                f"{fitness_from_cost(global_best_cost):.6f}"
            )

    # ------------------------------------------------------------
    # Verify allocation
    # ------------------------------------------------------------

    assert (
        global_best_allocation[0].sum()
        == conservancy_count
    )

    assert (
        global_best_allocation[1].sum()
        == supporting_count
    )

    assert np.all(
        global_best_allocation >= 0
    )

    return {
        "allocation": global_best_allocation,
        "best_cost": float(global_best_cost),
        "best_fitness": float(
            fitness_from_cost(global_best_cost)
        ),
        "best_position": global_best_position,
        "history": pd.DataFrame(history)
    }


# ================================================================
# 11. EXECUTE PSO
# ================================================================

pso_result = run_pso(
    conservancy_count=total_conservancy,
    supporting_count=total_supporting,
    sites=total_sites,
    weights=workload_weights,
    swarm_size=SWARM_SIZE,
    iterations=MAX_ITERATIONS,
    generator=rng
)

best_allocation = pso_result["allocation"]

pso_conservancy = best_allocation[0]
pso_supporting = best_allocation[1]

pso_total = (
    pso_conservancy + pso_supporting
)

best_cost = pso_result["best_cost"]
best_fitness = pso_result["best_fitness"]

history_df = pso_result["history"]

print("\nPSO Optimization Completed")
print("Best Cost:", best_cost)
print("Best Fitness:", best_fitness)

# ================================================================
# 12. SAVE SITE-WISE ALLOCATION CSV
# ================================================================

allocation_df = pd.DataFrame({
    "site_id": np.arange(1, total_sites + 1),
    "workload_weight": workload_weights,
    "target_conservancy": target_conservancy,
    "target_supporting": target_supporting,
    "pso_conservancy_workers": pso_conservancy,
    "pso_supporting_workers": pso_supporting,
    "pso_total_workers": pso_total
})

allocation_df["allocation_error"] = (
    np.abs(
        allocation_df["pso_conservancy_workers"]
        - allocation_df["target_conservancy"]
    )
    +
    np.abs(
        allocation_df["pso_supporting_workers"]
        - allocation_df["target_supporting"]
    )
)

allocation_df.to_csv(
    output_path("resource_allocation.csv"),
    index=False
)

print("\nPSO Site-Wise Allocation:")
print(allocation_df.head(10).to_string(index=False))

# ================================================================
# 13. SAVE CONVERGENCE CSV
# ================================================================

history_df.to_csv(
    output_path("convergence.csv"),
    index=False
)

# ================================================================
# 14. CONVERGENCE GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["iteration"],
    history_df["best_cost"],
    linewidth=2,
    label="Global Best Cost"
)

plt.plot(
    history_df["iteration"],
    history_df["mean_cost"],
    linewidth=2,
    alpha=0.7,
    label="Mean Swarm Cost"
)

plt.title("PSO Optimization Convergence")

plt.xlabel("Iteration")
plt.ylabel("Allocation Cost")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 15. PSO ACCURACY / FITNESS GRAPH
# ================================================================

# Fitness is an optimization metric, not ML accuracy.

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

axes[0].plot(
    history_df["iteration"],
    history_df["best_fitness"],
    linewidth=2
)

axes[0].set_title(
    "PSO Optimization Fitness"
)

axes[0].set_xlabel("Iteration")
axes[0].set_ylabel("Fitness")

axes[0].grid(True, alpha=0.3)

axes[1].axis("off")

axes[1].text(
    0.5,
    0.75,
    "ML Prediction Accuracy: N/A",
    ha="center",
    fontsize=15,
    fontweight="bold"
)

axes[1].text(
    0.5,
    0.48,
    f"Final PSO Fitness: {best_fitness:.6f}\n"
    f"Final Allocation Cost: {best_cost:.6f}",
    ha="center",
    fontsize=12
)

axes[1].text(
    0.5,
    0.2,
    "Fitness measures allocation objective quality.\n"
    "It is not prediction accuracy.",
    ha="center",
    fontsize=10
)

plt.tight_layout()

plt.savefig(
    output_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 16. PSO ALLOCATION HEATMAP
# ================================================================

heatmap_matrix = np.vstack([
    pso_conservancy,
    pso_supporting,
    pso_total
])

plt.figure(figsize=(16, 5))

sns.heatmap(
    heatmap_matrix,
    cmap="YlGnBu",
    yticklabels=[
        "Conservancy",
        "Supporting",
        "Total"
    ],
    xticklabels=np.arange(
        1, total_sites + 1
    ),
    cbar_kws={
        "label": "Allocated Workers"
    }
)

plt.title(
    "PSO Optimized Workforce Allocation Heatmap"
)

plt.xlabel("Illustrative Disposal Site")
plt.ylabel("Workforce Category")

plt.xticks(
    rotation=90,
    fontsize=6
)

plt.tight_layout()

plt.savefig(
    output_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 17. PSO VS EQUAL ALLOCATION
# ================================================================

equal_conservancy = equal_allocate(
    total_conservancy,
    total_sites
)

equal_supporting = equal_allocate(
    total_supporting,
    total_sites
)

equal_allocation = np.vstack([
    equal_conservancy,
    equal_supporting
])

equal_cost = allocation_cost(
    equal_allocation,
    target_conservancy,
    target_supporting
)

comparison_df = pd.DataFrame({
    "method": [
        "Equal Allocation",
        "PSO Optimization"
    ],

    "allocation_cost": [
        equal_cost,
        best_cost
    ],

    "optimization_fitness": [
        fitness_from_cost(equal_cost),
        best_fitness
    ],

    "total_allocated_workers": [
        int(equal_allocation.sum()),
        int(best_allocation.sum())
    ]
})

comparison_df.to_csv(
    output_path("comparison.csv"),
    index=False
)

plt.figure(figsize=(9, 6))

bars = plt.bar(
    comparison_df["method"],
    comparison_df["allocation_cost"]
)

plt.title(
    "PSO vs Equal Allocation\n"
    "Hypothetical Workload Scenario"
)

plt.ylabel("Mean Absolute Allocation Deviation")

for bar in bars:

    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.savefig(
    output_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 18. GENERATE PSO RESULT CSV
# ================================================================

result_df = pd.DataFrame([{
    "dataset_year": "2017-18",
    "algorithm": "Particle Swarm Optimization",
    "total_conservancy_workers": total_conservancy,
    "total_supporting_workers": total_supporting,
    "total_workers": total_workers,
    "total_disposal_sites": total_sites,
    "allocated_conservancy_workers": int(
        pso_conservancy.sum()
    ),
    "allocated_supporting_workers": int(
        pso_supporting.sum()
    ),
    "allocated_total_workers": int(
        pso_total.sum()
    ),
    "minimum_workers_per_site": int(
        pso_total.min()
    ),
    "maximum_workers_per_site": int(
        pso_total.max()
    ),
    "average_workers_per_site": float(
        pso_total.mean()
    ),
    "pso_best_cost": float(best_cost),
    "pso_best_fitness": float(best_fitness),
    "equal_allocation_cost": float(equal_cost),
    "improvement_over_equal_cost": float(
        equal_cost - best_cost
    ),
    "prediction_accuracy": np.nan,
    "evaluation_note": (
        "PSO fitness is not ML prediction accuracy"
    )
}])

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nPSO RESULTS:")
print(result_df.to_string(index=False))

# ================================================================
# 19. RESULT GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 6)
)

categories = [
    "Conservancy",
    "Supporting",
    "Total"
]

values = [
    int(pso_conservancy.sum()),
    int(pso_supporting.sum()),
    int(pso_total.sum())
]

bars = axes[0].bar(
    categories,
    values
)

axes[0].set_title(
    "PSO Workforce Allocation Totals"
)

axes[0].set_ylabel("Number of Workers")

for bar in bars:

    h = bar.get_height()

    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        h,
        str(int(h)),
        ha="center",
        va="bottom"
    )

axes[1].plot(
    allocation_df["site_id"],
    allocation_df["pso_total_workers"],
    label="PSO Allocated Workers"
)

axes[1].plot(
    allocation_df["site_id"],
    (
        target_conservancy
        + target_supporting
    ),
    linestyle="--",
    label="Hypothetical Workload Target"
)

axes[1].set_title(
    "PSO Site-Wise Allocation"
)

axes[1].set_xlabel("Site ID")
axes[1].set_ylabel("Workers")

axes[1].legend()

plt.tight_layout()

plt.savefig(
    output_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 20. FUTURE SCENARIO OPTIMIZATION
# ================================================================

# These are illustrative projections.
#
# Workforce growth: 4% per year
# Disposal site growth: 3% per year
#
# PSO optimizes allocations for each scenario.
# It does not learn future demand from historical data.

prediction_rows = []

for year in range(
    BASE_YEAR,
    BASE_YEAR + FORECAST_YEARS + 1
):

    t = year - BASE_YEAR

    projected_sites = max(
        1,
        round(
            total_sites
            * (1 + SITE_GROWTH_RATE) ** t
        )
    )

    projected_conservancy = round(
        total_conservancy
        * (1 + WORKFORCE_GROWTH_RATE) ** t
    )

    projected_supporting = round(
        total_supporting
        * (1 + WORKFORCE_GROWTH_RATE) ** t
    )

    projected_weights = np.linspace(
        1.0,
        2.0,
        projected_sites
    )

    projected_weights = (
        projected_weights
        / projected_weights.sum()
    )

    # Reduced iterations for faster scenario processing.
    future_result = run_pso(
        conservancy_count=projected_conservancy,
        supporting_count=projected_supporting,
        sites=projected_sites,
        weights=projected_weights,
        swarm_size=20,
        iterations=60,
        generator=rng
    )

    future_allocation = future_result["allocation"]

    prediction_rows.append({
        "year": year,
        "projected_disposal_sites": projected_sites,
        "projected_conservancy_workers": (
            projected_conservancy
        ),
        "projected_supporting_workers": (
            projected_supporting
        ),
        "projected_total_workers": (
            projected_conservancy
            + projected_supporting
        ),
        "pso_allocated_conservancy": int(
            future_allocation[0].sum()
        ),
        "pso_allocated_supporting": int(
            future_allocation[1].sum()
        ),
        "pso_allocation_cost": float(
            future_result["best_cost"]
        ),
        "pso_optimization_fitness": float(
            future_result["best_fitness"]
        ),
        "projection_type": (
            "Assumption-based scenario"
        )
    })

prediction_df = pd.DataFrame(
    prediction_rows
)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nPSO FUTURE SCENARIOS:")
print(prediction_df.to_string(index=False))

# ================================================================
# 21. PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(12, 7))

plt.plot(
    prediction_df["year"],
    prediction_df["projected_conservancy_workers"],
    marker="o",
    linewidth=2,
    label="Conservancy Workers"
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_supporting_workers"],
    marker="s",
    linewidth=2,
    label="Supporting Workers"
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_total_workers"],
    marker="^",
    linewidth=2,
    label="Total Workforce"
)

plt.title(
    "PSO Scenario-Based Workforce Projections\n"
    "4% Annual Workforce Growth Assumption"
)

plt.xlabel("Year")
plt.ylabel("Projected Workers")

plt.xticks(
    prediction_df["year"],
    rotation=45
)

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 22. SAVE PSO MODEL AS PKL
# ================================================================

model_package = {
    "algorithm": "Particle Swarm Optimization",
    "best_position": pso_result["best_position"],
    "best_allocation": best_allocation,
    "best_cost": best_cost,
    "best_fitness": best_fitness,
    "workload_weights": workload_weights,
    "allocation": allocation_df,
    "convergence": history_df,
    "comparison": comparison_df,
    "prediction": prediction_df,
    "is_trained_predictive_model": False
}

with open(
    output_path("model.pkl"),
    "wb"
) as file:

    pickle.dump(
        model_package,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print("\nPSO PKL Model Saved.")

# ================================================================
# 23. SAVE PSO MODEL AS H5
# ================================================================

with h5py.File(
    output_path("model.h5"),
    "w"
) as h5:

    h5.attrs["algorithm"] = (
        "Particle Swarm Optimization"
    )

    h5.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    h5.attrs["is_predictive_ml_model"] = False

    h5.attrs["best_cost"] = float(best_cost)
    h5.attrs["best_fitness"] = float(best_fitness)

    h5.create_dataset(
        "best_position",
        data=pso_result["best_position"]
    )

    h5.create_dataset(
        "best_allocation",
        data=best_allocation
    )

    h5.create_dataset(
        "workload_weights",
        data=workload_weights
    )

    h5.create_dataset(
        "iteration",
        data=history_df["iteration"].to_numpy()
    )

    h5.create_dataset(
        "best_cost_history",
        data=history_df["best_cost"].to_numpy()
    )

    h5.create_dataset(
        "best_fitness_history",
        data=history_df["best_fitness"].to_numpy()
    )

    h5.create_dataset(
        "pso_total_workers_by_site",
        data=pso_total
    )

print("PSO H5 File Saved.")

# ================================================================
# 24. SAVE YAML CONFIGURATION
# ================================================================

config = {
    "project": {
        "name": (
            "AI-Based Conservancy Workforce Analysis, "
            "Waste Disposal Infrastructure Assessment, "
            "and Resource Optimization System"
        ),
        "algorithm": "Particle Swarm Optimization",
        "dataset_year": "2017-18",
        "created_at": datetime.now().isoformat()
    },

    "dataset": {
        "path": str(CSV_PATH),
        "records": int(len(df)),
        "total_conservancy_workers": total_conservancy,
        "total_supporting_workers": total_supporting,
        "total_disposal_sites": total_sites
    },

    "pso_parameters": {
        "swarm_size": SWARM_SIZE,
        "iterations": MAX_ITERATIONS,
        "inertia_start": INERTIA_START,
        "inertia_end": INERTIA_END,
        "cognitive_coefficient": (
            COGNITIVE_COEFFICIENT
        ),
        "social_coefficient": (
            SOCIAL_COEFFICIENT
        ),
        "velocity_limit": VELOCITY_LIMIT,
        "random_seed": SEED
    },

    "optimization": {
        "objective": (
            "Minimize mean absolute deviation from "
            "hypothetical workload-based staffing targets"
        ),
        "best_cost": float(best_cost),
        "best_fitness": float(best_fitness),
        "workload_assumption": (
            "Illustrative site weights increase from 1 to 2"
        )
    },

    "forecast": {
        "base_year": BASE_YEAR,
        "forecast_years": FORECAST_YEARS,
        "workforce_growth_rate": (
            WORKFORCE_GROWTH_RATE
        ),
        "site_growth_rate": SITE_GROWTH_RATE,
        "type": "Assumption-based scenario"
    }
}

with open(
    output_path("model_config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("PSO YAML Configuration Saved.")

# ================================================================
# 25. SAVE JSON METADATA
# ================================================================

generated_outputs = [
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_resource_allocation.csv",
    "pso_comparison.csv",
    "pso_convergence.csv",
    "pso_convergence_graph.png",
    "pso_model.pkl",
    "pso_model.h5",
    "pso_model_config.yaml",
    "pso_project_metadata.json"
]

metadata = {
    "project_name": (
        "AI-Based Conservancy Workforce Analysis, "
        "Waste Disposal Infrastructure Assessment, "
        "and Resource Optimization System"
    ),

    "algorithm": "Particle Swarm Optimization",

    "created_at": datetime.now().isoformat(),

    "dataset": {
        "year": "2017-18",
        "records": int(len(df)),
        "conservancy_workers": total_conservancy,
        "supporting_workers": total_supporting,
        "disposal_sites": total_sites,
        "total_workers": total_workers
    },

    "optimization_results": {
        "best_cost": float(best_cost),
        "best_fitness": float(best_fitness),
        "equal_allocation_cost": float(equal_cost),
        "improvement_over_equal": float(
            equal_cost - best_cost
        )
    },

    "model_evaluation": {
        "prediction_accuracy": None,
        "r2_score": None,
        "mae": None,
        "rmse": None,
        "reason": (
            "Insufficient historical observations "
            "for predictive ML evaluation"
        )
    },

    "forecast_assumptions": {
        "workforce_growth_rate": (
            WORKFORCE_GROWTH_RATE
        ),
        "site_growth_rate": SITE_GROWTH_RATE
    },

    "generated_outputs": generated_outputs
}

with open(
    output_path("project_metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("PSO JSON Metadata Saved.")

# ================================================================
# 26. VERIFY ALL OUTPUT FILES
# ================================================================

print("\n" + "=" * 70)
print("PSO OUTPUT FILE VERIFICATION")
print("=" * 70)

for filename in generated_outputs:

    path = BASE_DIR / filename

    if path.exists():

        print(
            f"[SUCCESS] {filename} "
            f"({path.stat().st_size:,} bytes)"
        )

    else:

        print(f"[FAILED] {filename}")

print("\nAll PSO results saved to:")
print(BASE_DIR)

print("\nPSO PROJECT EXECUTION COMPLETED SUCCESSFULLY!")
```
