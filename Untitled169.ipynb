```python
import json
import pickle
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore", category=FutureWarning)

sns.set_theme(style="whitegrid")

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Conservancy Workforce Analysis, Waste Disposal Infrastructure Assessment, and Resource Optimization System"
)

CSV_PATH = BASE_DIR / (
    "number_of conservancy_workers_engaged_in_conservancy_"
    "garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

POPULATION_SIZE = 35
GENERATIONS = 150
ELITE_COUNT = 8
CLONES_PER_ELITE = 5
RANDOM_IMMIGRANTS = 5

BASE_YEAR = 2018
FORECAST_YEARS = 10
WORKFORCE_GROWTH_RATE = 0.04
SITE_GROWTH_RATE = 0.03

print("=" * 70)
print("ARTIFICIAL IMMUNE SYSTEM - CONSERVANCY OPTIMIZATION")
print("=" * 70)

# ================================================================
# 2. LOAD AND VALIDATE DATA
# ================================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

COLUMN_MAPPING = {
    "Number of workers engaged in conservancy work":
        "conservancy_workers",

    "Number of garbage and sewage disposal dump site":
        "disposal_sites",

    "Number of persons engaged in other connected works related with conservancy":
        "supporting_workers"
}

missing = [
    c for c in COLUMN_MAPPING
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}\n"
        f"Available: {list(df.columns)}"
    )

data = df.rename(columns=COLUMN_MAPPING).copy()

required = [
    "conservancy_workers",
    "supporting_workers",
    "disposal_sites"
]

for col in required:
    data[col] = pd.to_numeric(
        data[col], errors="coerce"
    )

if data[required].isna().any().any():
    raise ValueError("Invalid or missing numeric values.")

if (data[required] < 0).any().any():
    raise ValueError("Negative counts are invalid.")

values = data[required].to_numpy()

if not np.all(values == np.floor(values)):
    raise ValueError("Counts must be integers.")

data[required] = data[required].astype(int)

conservancy = int(data["conservancy_workers"].sum())
supporting = int(data["supporting_workers"].sum())
sites = int(data["disposal_sites"].sum())

if sites <= 0:
    raise ValueError("At least one disposal site is required.")

total_workers = conservancy + supporting

print("\nDataset:", df.shape)
print("Conservancy Workers:", conservancy)
print("Supporting Workers:", supporting)
print("Disposal Sites:", sites)
print("Total Workers:", total_workers)

# ================================================================
# 3. WORKLOAD SCENARIO
# ================================================================

# Actual site-level workload is not provided.
#
# These illustrative weights gradually increase from 1 to 2.
# They do NOT represent observed site workloads.
#
# Replace these weights with measured site workload data
# when available.

workload_weights = np.linspace(1.0, 2.0, sites)

workload_weights = (
    workload_weights / workload_weights.sum()
)

target_conservancy = (
    conservancy * workload_weights
)

target_supporting = (
    supporting * workload_weights
)

# ================================================================
# 4. ALLOCATION UTILITIES
# ================================================================

def proportional_allocate(total, weights):
    """
    Integer allocation using largest remainder method.
    Preserves exact workforce total.
    """

    weights = np.asarray(weights, dtype=float)
    weights = weights / weights.sum()

    raw = total * weights

    result = np.floor(raw).astype(int)

    remaining = int(total - result.sum())

    fractions = raw - result

    order = np.argsort(-fractions)

    result[order[:remaining]] += 1

    return result


def equal_allocate(total, n):
    """
    Equal integer distribution.
    """

    base, remainder = divmod(total, n)

    result = np.full(n, base, dtype=int)
    result[:remainder] += 1

    return result


def random_allocate(total, n, rng):
    """
    Random nonnegative integer allocation
    preserving the exact total.
    """

    if total == 0:
        return np.zeros(n, dtype=int)

    probabilities = rng.dirichlet(
        np.ones(n)
    )

    return rng.multinomial(
        total, probabilities
    )


# ================================================================
# 5. AIS FITNESS FUNCTION
# ================================================================

def allocation_cost(candidate):
    """
    Candidate shape: (2, sites)

    Row 0: Conservancy allocation
    Row 1: Supporting allocation

    Minimize mean absolute deviation from
    hypothetical workload-based targets.
    """

    c = candidate[0]
    s = candidate[1]

    error_c = np.mean(
        np.abs(c - target_conservancy)
    )

    error_s = np.mean(
        np.abs(s - target_supporting)
    )

    return float(error_c + error_s)


def fitness(candidate):
    """
    Higher fitness is better.
    """

    return 1.0 / (
        1.0 + allocation_cost(candidate)
    )


# ================================================================
# 6. INITIALIZE ANTIBODY POPULATION
# ================================================================

def create_antibody():
    """
    Antibody represents a feasible workforce allocation.
    """

    return np.vstack([
        random_allocate(conservancy, sites, rng),
        random_allocate(supporting, sites, rng)
    ])


def initialize_population():

    population = []

    # Equal distribution baseline
    population.append(
        np.vstack([
            equal_allocate(conservancy, sites),
            equal_allocate(supporting, sites)
        ])
    )

    # Workload-proportional baseline
    population.append(
        np.vstack([
            proportional_allocate(
                conservancy, workload_weights
            ),
            proportional_allocate(
                supporting, workload_weights
            )
        ])
    )

    while len(population) < POPULATION_SIZE:
        population.append(create_antibody())

    return population


# ================================================================
# 7. SOMATIC HYPERMUTATION
# ================================================================

def mutate_antibody(
    antibody,
    mutation_strength=1.0
):
    """
    Transfer workers between randomly selected sites.

    Total conservancy and supporting workers are preserved.
    No negative allocations are allowed.
    """

    mutated = antibody.copy()

    mutation_steps = max(
        1,
        int(sites * mutation_strength)
    )

    for _ in range(mutation_steps):

        worker_type = int(
            rng.integers(0, 2)
        )

        donor = int(
            rng.integers(0, sites)
        )

        receiver = int(
            rng.integers(0, sites)
        )

        if donor == receiver:
            continue

        available = int(
            mutated[worker_type, donor]
        )

        if available <= 0:
            continue

        amount = int(
            rng.integers(
                1,
                min(available, 3) + 1
            )
        )

        mutated[worker_type, donor] -= amount
        mutated[worker_type, receiver] += amount

    return mutated


# ================================================================
# 8. AIS CLONAL SELECTION OPTIMIZATION
# ================================================================

population = initialize_population()

history = []

best_antibody = None
best_cost = float("inf")

for generation in range(GENERATIONS):

    population.sort(key=allocation_cost)

    current_best = population[0]
    current_cost = allocation_cost(current_best)

    if current_cost < best_cost:

        best_cost = current_cost

        best_antibody = current_best.copy()

    elites = [
        antibody.copy()
        for antibody in population[:ELITE_COUNT]
    ]

    clones = []

    for rank, antibody in enumerate(elites):

        # Better antibodies receive weaker mutation.
        mutation_strength = (
            0.02 + 0.25 * (rank + 1) / ELITE_COUNT
        )

        for _ in range(CLONES_PER_ELITE):

            clone = mutate_antibody(
                antibody,
                mutation_strength
            )

            clones.append(clone)

    combined = (
        population
        + clones
        + [create_antibody()
           for _ in range(RANDOM_IMMIGRANTS)]
    )

    combined.sort(key=allocation_cost)

    # Keep top antibodies and a global-best memory.
    population = [
        best_antibody.copy()
    ]

    for antibody in combined:

        if len(population) >= POPULATION_SIZE:
            break

        population.append(antibody.copy())

    population.sort(key=allocation_cost)

    generation_best_cost = allocation_cost(
        population[0]
    )

    generation_mean_cost = float(
        np.mean([
            allocation_cost(a)
            for a in population
        ])
    )

    history.append({
        "generation": generation + 1,
        "best_cost": generation_best_cost,
        "mean_cost": generation_mean_cost,
        "best_fitness": (
            1 / (1 + generation_best_cost)
        )
    })

    if (
        (generation + 1) % 10 == 0
        or generation == 0
    ):
        print(
            f"Generation {generation + 1:3d} | "
            f"Cost: {generation_best_cost:.6f} | "
            f"Fitness: {fitness(population[0]):.6f}"
        )

# Update global best from final population.
final_candidate = min(
    population, key=allocation_cost
)

if allocation_cost(final_candidate) < best_cost:

    best_antibody = final_candidate.copy()
    best_cost = allocation_cost(final_candidate)

ais_conservancy = best_antibody[0]
ais_supporting = best_antibody[1]

ais_total = (
    ais_conservancy + ais_supporting
)

assert ais_conservancy.sum() == conservancy
assert ais_supporting.sum() == supporting
assert np.all(best_antibody >= 0)

print("\nAIS Optimization Completed!")
print("Best Cost:", best_cost)
print("Best Fitness:", fitness(best_antibody))

# ================================================================
# 9. SAVE SITE-WISE AIS ALLOCATION
# ================================================================

allocation_df = pd.DataFrame({
    "site_id": np.arange(1, sites + 1),
    "workload_weight": workload_weights,
    "target_conservancy": target_conservancy,
    "target_supporting": target_supporting,
    "ais_conservancy_workers": ais_conservancy,
    "ais_supporting_workers": ais_supporting,
    "ais_total_workers": ais_total
})

allocation_df["allocation_error"] = (
    np.abs(
        allocation_df["ais_conservancy_workers"]
        - allocation_df["target_conservancy"]
    )
    +
    np.abs(
        allocation_df["ais_supporting_workers"]
        - allocation_df["target_supporting"]
    )
)

allocation_df.to_csv(
    BASE_DIR / "ais_resource_allocation.csv",
    index=False
)

print("\nAIS Allocation:")
print(allocation_df.head(10).to_string(index=False))

# ================================================================
# 10. SAVE CONVERGENCE HISTORY
# ================================================================

history_df = pd.DataFrame(history)

history_df.to_csv(
    BASE_DIR / "ais_convergence.csv",
    index=False
)

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["generation"],
    history_df["best_cost"],
    label="Best Cost",
    linewidth=2
)

plt.plot(
    history_df["generation"],
    history_df["mean_cost"],
    label="Population Mean Cost",
    linewidth=2,
    alpha=0.7
)

plt.title("AIS Optimization Convergence")
plt.xlabel("Generation")
plt.ylabel("Allocation Cost")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 11. AIS ACCURACY GRAPH
# ================================================================

# This graph shows optimization fitness.
# It is NOT predictive ML accuracy.

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

axes[0].plot(
    history_df["generation"],
    history_df["best_fitness"],
    linewidth=2
)

axes[0].set_title(
    "AIS Best Fitness by Generation"
)

axes[0].set_xlabel("Generation")
axes[0].set_ylabel("Optimization Fitness")
axes[0].grid(True, alpha=0.3)

axes[1].axis("off")

axes[1].text(
    0.5, 0.75,
    "ML Prediction Accuracy: N/A",
    ha="center",
    fontsize=15,
    fontweight="bold"
)

axes[1].text(
    0.5, 0.48,
    f"Final AIS Fitness: {fitness(best_antibody):.6f}\n"
    f"Final Allocation Cost: {best_cost:.6f}",
    ha="center",
    fontsize=12
)

axes[1].text(
    0.5, 0.2,
    "Fitness measures optimization objective quality,\n"
    "not prediction accuracy.",
    ha="center",
    fontsize=10
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 12. AIS HEATMAP
# ================================================================

heatmap_data = np.vstack([
    ais_conservancy,
    ais_supporting,
    ais_total
])

plt.figure(figsize=(16, 5))

sns.heatmap(
    heatmap_data,
    cmap="YlGnBu",
    yticklabels=[
        "Conservancy",
        "Supporting",
        "Total"
    ],
    xticklabels=np.arange(1, sites + 1),
    cbar_kws={"label": "Allocated Workers"}
)

plt.title(
    "AIS Optimized Workforce Allocation Heatmap"
)

plt.xlabel("Illustrative Disposal Site")
plt.ylabel("Workforce Category")

plt.xticks(
    rotation=90,
    fontsize=6
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 13. AIS VS EQUAL ALLOCATION COMPARISON
# ================================================================

equal_conservancy = equal_allocate(
    conservancy, sites
)

equal_supporting = equal_allocate(
    supporting, sites
)

equal_antibody = np.vstack([
    equal_conservancy,
    equal_supporting
])

equal_cost = allocation_cost(equal_antibody)
ais_cost = allocation_cost(best_antibody)

comparison_df = pd.DataFrame({
    "method": [
        "Equal Allocation",
        "AIS Optimization"
    ],
    "allocation_cost": [
        equal_cost,
        ais_cost
    ],
    "optimization_fitness": [
        fitness(equal_antibody),
        fitness(best_antibody)
    ],
    "total_allocated_workers": [
        int(equal_antibody.sum()),
        int(best_antibody.sum())
    ]
})

comparison_df.to_csv(
    BASE_DIR / "ais_comparison.csv",
    index=False
)

plt.figure(figsize=(9, 6))

bars = plt.bar(
    comparison_df["method"],
    comparison_df["allocation_cost"]
)

plt.title(
    "AIS vs Equal Allocation\n"
    "Hypothetical Workload Scenario"
)

plt.ylabel(
    "Mean Absolute Allocation Deviation"
)

for bar in bars:

    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 14. GENERATE AIS RESULT CSV
# ================================================================

result_df = pd.DataFrame([{
    "dataset_year": "2017-18",
    "algorithm": "AIS Clonal Selection",
    "total_conservancy_workers": conservancy,
    "total_supporting_workers": supporting,
    "total_workers": total_workers,
    "total_disposal_sites": sites,
    "allocated_conservancy_workers": int(
        ais_conservancy.sum()
    ),
    "allocated_supporting_workers": int(
        ais_supporting.sum()
    ),
    "allocated_total_workers": int(
        ais_total.sum()
    ),
    "minimum_workers_per_site": int(
        ais_total.min()
    ),
    "maximum_workers_per_site": int(
        ais_total.max()
    ),
    "average_workers_per_site": float(
        ais_total.mean()
    ),
    "ais_best_cost": float(best_cost),
    "ais_best_fitness": float(
        fitness(best_antibody)
    ),
    "equal_allocation_cost": float(equal_cost),
    "improvement_over_equal_cost": float(
        equal_cost - ais_cost
    ),
    "prediction_accuracy": np.nan,
    "evaluation_note": (
        "AIS fitness is not ML prediction accuracy"
    )
}])

result_df.to_csv(
    BASE_DIR / "ais_result.csv",
    index=False
)

print("\nAIS RESULT:")
print(result_df.to_string(index=False))

# ================================================================
# 15. AIS RESULT GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 6)
)

categories = [
    "Conservancy",
    "Supporting",
    "Total"
]

values = [
    int(ais_conservancy.sum()),
    int(ais_supporting.sum()),
    int(ais_total.sum())
]

bars = axes[0].bar(
    categories,
    values
)

axes[0].set_title(
    "AIS Workforce Allocation Totals"
)

axes[0].set_ylabel("Number of Workers")

for bar in bars:

    h = bar.get_height()

    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        h,
        str(int(h)),
        ha="center",
        va="bottom"
    )

axes[1].plot(
    allocation_df["site_id"],
    allocation_df["ais_total_workers"],
    label="AIS Allocated Workers"
)

axes[1].plot(
    allocation_df["site_id"],
    (
        target_conservancy
        + target_supporting
    ),
    linestyle="--",
    label="Hypothetical Workload Target"
)

axes[1].set_title(
    "AIS Site-Wise Allocation"
)

axes[1].set_xlabel("Site ID")
axes[1].set_ylabel("Workers")
axes[1].legend()

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 16. AIS SCENARIO-BASED FUTURE PROJECTIONS
# ================================================================

# These are scenario estimates, not predictions learned
# from historical data.
#
# AIS optimizes the allocation for each projected year.
# The 4% workforce and 3% site growth rates are assumptions.

def run_ais_for_projection(
    projected_conservancy,
    projected_supporting,
    projected_sites
):
    """
    Lightweight AIS optimization for a future scenario.

    Uses the same clonal-selection approach, with
    fewer generations for faster execution.
    """

    weights = np.linspace(
        1.0, 2.0, projected_sites
    )

    weights = weights / weights.sum()

    target_c = projected_conservancy * weights
    target_s = projected_supporting * weights

    def local_cost(a):

        return float(
            np.mean(np.abs(a[0] - target_c))
            +
            np.mean(np.abs(a[1] - target_s))
        )

    def make_candidate():

        return np.vstack([
            random_allocate(
                projected_conservancy,
                projected_sites,
                rng
            ),
            random_allocate(
                projected_supporting,
                projected_sites,
                rng
            )
        ])

    population_local = [
        np.vstack([
            proportional_allocate(
                projected_conservancy,
                weights
            ),
            proportional_allocate(
                projected_supporting,
                weights
            )
        ]),
        np.vstack([
            equal_allocate(
                projected_conservancy,
                projected_sites
            ),
            equal_allocate(
                projected_supporting,
                projected_sites
            )
        ])
    ]

    while len(population_local) < 20:
        population_local.append(make_candidate())

    for _ in range(60):

        population_local.sort(key=local_cost)

        elites_local = population_local[:5]

        clones_local = []

        for rank, elite in enumerate(elites_local):

            for _ in range(4):

                clone = elite.copy()

                steps = max(
                    1,
                    int(
                        projected_sites
                        * (0.02 + 0.03 * rank)
                    )
                )

                for _ in range(steps):

                    row = int(rng.integers(0, 2))
                    donor = int(
                        rng.integers(0, projected_sites)
                    )
                    receiver = int(
                        rng.integers(0, projected_sites)
                    )

                    if (
                        donor != receiver
                        and clone[row, donor] > 0
                    ):
                        clone[row, donor] -= 1
                        clone[row, receiver] += 1

                clones_local.append(clone)

        population_local = sorted(
            population_local + clones_local,
            key=local_cost
        )[:20]

    best = min(
        population_local,
        key=local_cost
    )

    return {
        "allocation_cost": local_cost(best),
        "fitness": 1 / (1 + local_cost(best)),
        "allocated_conservancy": int(best[0].sum()),
        "allocated_supporting": int(best[1].sum())
    }


prediction_rows = []

for year in range(
    BASE_YEAR,
    BASE_YEAR + FORECAST_YEARS + 1
):

    t = year - BASE_YEAR

    projected_sites = max(
        1,
        round(
            sites * (1 + SITE_GROWTH_RATE) ** t
        )
    )

    projected_conservancy = round(
        conservancy
        * (1 + WORKFORCE_GROWTH_RATE) ** t
    )

    projected_supporting = round(
        supporting
        * (1 + WORKFORCE_GROWTH_RATE) ** t
    )

    optimized = run_ais_for_projection(
        projected_conservancy,
        projected_supporting,
        projected_sites
    )

    prediction_rows.append({
        "year": year,
        "projected_disposal_sites": projected_sites,
        "projected_conservancy_workers": (
            projected_conservancy
        ),
        "projected_supporting_workers": (
            projected_supporting
        ),
        "projected_total_workers": (
            projected_conservancy
            + projected_supporting
        ),
        "ais_allocated_conservancy": (
            optimized["allocated_conservancy"]
        ),
        "ais_allocated_supporting": (
            optimized["allocated_supporting"]
        ),
        "ais_allocation_cost": (
            optimized["allocation_cost"]
        ),
        "ais_optimization_fitness": (
            optimized["fitness"]
        ),
        "projection_type": (
            "Assumption-based scenario"
        )
    })

prediction_df = pd.DataFrame(
    prediction_rows
)

prediction_df.to_csv(
    BASE_DIR / "ais_prediction.csv",
    index=False
)

print("\nAIS FUTURE SCENARIOS:")
print(prediction_df.to_string(index=False))

# ================================================================
# 17. AIS PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(12, 7))

plt.plot(
    prediction_df["year"],
    prediction_df["projected_conservancy_workers"],
    marker="o",
    label="Conservancy Workers",
    linewidth=2
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_supporting_workers"],
    marker="s",
    label="Supporting Workers",
    linewidth=2
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_total_workers"],
    marker="^",
    label="Total Workforce",
    linewidth=2
)

plt.title(
    "AIS Scenario-Based Workforce Projections\n"
    "4% Annual Workforce Growth Assumption"
)

plt.xlabel("Year")
plt.ylabel("Projected Workers")

plt.xticks(
    prediction_df["year"],
    rotation=45
)

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 18. SAVE AIS PKL MODEL
# ================================================================

model_package = {
    "algorithm": "AIS Clonal Selection",
    "best_antibody": best_antibody,
    "best_cost": best_cost,
    "best_fitness": fitness(best_antibody),
    "workload_weights": workload_weights,
    "allocation": allocation_df,
    "convergence": history_df,
    "comparison": comparison_df,
    "prediction": prediction_df,
    "is_trained_predictive_model": False
}

with open(
    BASE_DIR / "ais_model.pkl",
    "wb"
) as file:

    pickle.dump(
        model_package,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

# ================================================================
# 19. SAVE AIS H5 FILE
# ================================================================

with h5py.File(
    BASE_DIR / "ais_model.h5",
    "w"
) as h5:

    h5.attrs["algorithm"] = (
        "Artificial Immune System - Clonal Selection"
    )

    h5.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    h5.attrs["is_predictive_ml_model"] = False

    h5.attrs["best_cost"] = float(best_cost)

    h5.attrs["best_fitness"] = float(
        fitness(best_antibody)
    )

    h5.create_dataset(
        "best_antibody",
        data=best_antibody
    )

    h5.create_dataset(
        "workload_weights",
        data=workload_weights
    )

    h5.create_dataset(
        "generation",
        data=history_df["generation"].to_numpy()
    )

    h5.create_dataset(
        "best_cost_history",
        data=history_df["best_cost"].to_numpy()
    )

    h5.create_dataset(
        "fitness_history",
        data=history_df["best_fitness"].to_numpy()
    )

    h5.create_dataset(
        "ais_total_workers_by_site",
        data=ais_total
    )

# ================================================================
# 20. SAVE AIS YAML CONFIGURATION
# ================================================================

config = {
    "project": {
        "name": (
            "AI-Based Conservancy Workforce Analysis, "
            "Waste Disposal Infrastructure Assessment, "
            "and Resource Optimization System"
        ),
        "algorithm": "AIS Clonal Selection",
        "dataset_year": "2017-18",
        "created_at": datetime.now().isoformat()
    },

    "dataset": {
        "path": str(CSV_PATH),
        "records": int(len(df)),
        "total_conservancy_workers": conservancy,
        "total_supporting_workers": supporting,
        "total_disposal_sites": sites
    },

    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "elite_count": ELITE_COUNT,
        "clones_per_elite": CLONES_PER_ELITE,
        "random_immigrants": RANDOM_IMMIGRANTS,
        "random_seed": SEED
    },

    "optimization": {
        "objective": (
            "Minimize mean absolute deviation from "
            "hypothetical workload-based staffing targets"
        ),
        "best_cost": float(best_cost),
        "best_fitness": float(
            fitness(best_antibody)
        ),
        "workload_assumption": (
            "Illustrative weights increase from 1 to 2"
        )
    },

    "forecast": {
        "base_year": BASE_YEAR,
        "forecast_years": FORECAST_YEARS,
        "workforce_growth_rate": (
            WORKFORCE_GROWTH_RATE
        ),
        "site_growth_rate": SITE_GROWTH_RATE,
        "type": "Assumption-based scenario"
    }
}

with open(
    BASE_DIR / "ais_model_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ================================================================
# 21. SAVE AIS JSON METADATA
# ================================================================

metadata = {
    "project_name": (
        "AI-Based Conservancy Workforce Analysis, "
        "Waste Disposal Infrastructure Assessment, "
        "and Resource Optimization System"
    ),

    "algorithm": "Artificial Immune System",

    "variant": "Clonal Selection Algorithm",

    "created_at": datetime.now().isoformat(),

    "dataset": {
        "year": "2017-18",
        "records": int(len(df)),
        "conservancy_workers": conservancy,
        "supporting_workers": supporting,
        "disposal_sites": sites,
        "total_workers": total_workers
    },

    "optimization_results": {
        "best_cost": float(best_cost),
        "best_fitness": float(
            fitness(best_antibody)
        ),
        "equal_allocation_cost": float(equal_cost),
        "improvement_over_equal": float(
            equal_cost - best_cost
        )
    },

    "model_evaluation": {
        "prediction_accuracy": None,
        "r2_score": None,
        "mae": None,
        "rmse": None,
        "reason": (
            "Insufficient historical observations "
            "for predictive ML evaluation"
        )
    },

    "forecast_assumptions": {
        "workforce_growth_rate": (
            WORKFORCE_GROWTH_RATE
        ),
        "site_growth_rate": SITE_GROWTH_RATE
    },

    "generated_outputs": [
        "ais_accuracy_graph.png",
        "ais_heatmap.png",
        "ais_comparison_graph.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_resource_allocation.csv",
        "ais_comparison.csv",
        "ais_convergence.csv",
        "ais_convergence_graph.png",
        "ais_model.pkl",
        "ais_model.h5",
        "ais_model_config.yaml",
        "ais_project_metadata.json"
    ]
}

with open(
    BASE_DIR / "ais_project_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

# ================================================================
# 22. VERIFY GENERATED FILES
# ================================================================

files = metadata["generated_outputs"]

print("\n" + "=" * 70)
print("AIS OUTPUT FILE VERIFICATION")
print("=" * 70)

for filename in files:

    path = BASE_DIR / filename

    if path.exists():

        print(
            f"[SUCCESS] {filename} "
            f"({path.stat().st_size:,} bytes)"
        )

    else:

        print(f"[FAILED] {filename}")

print("\nAll AIS results saved to:")
print(BASE_DIR)

print("\nAIS PROJECT EXECUTION COMPLETED!")
```
