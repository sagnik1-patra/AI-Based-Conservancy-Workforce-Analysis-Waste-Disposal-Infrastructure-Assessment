```python
import os
import json
import pickle
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import h5py
import yaml
import matplotlib.pyplot as plt


# ============================================================
# 1. PROJECT PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Conservancy Workforce Analysis, Waste Disposal Infrastructure Assessment, and Resource Optimization System"
)

CSV_PATH = BASE_DIR / (
    "number_of conservancy_workers_engaged_in_conservancy_"
    "garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = BASE_DIR / "conservancy_workforce_model.h5"
PKL_PATH = BASE_DIR / "conservancy_workforce_model.pkl"
YAML_PATH = BASE_DIR / "model_config.yaml"
JSON_PATH = BASE_DIR / "project_metadata.json"

print("=" * 65)
print("CONSERVANCY WORKFORCE ANALYSIS SYSTEM")
print("=" * 65)

print("\nProject Directory:", BASE_DIR)


# ============================================================
# 2. LOAD DATASET
# ============================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nDataset Loaded Successfully!")
print("\nDataset Shape:", df.shape)
print("\nDataset Preview:")
print(df.head())

print("\nDataset Information:")
df.info()


# ============================================================
# 3. DATA VALIDATION AND PREPROCESSING
# ============================================================

COLUMN_MAPPING = {
    "Number of workers engaged in conservancy work":
        "conservancy_workers",

    "Number of garbage and sewage disposal dump site":
        "disposal_sites",

    "Number of persons engaged in other connected works related with conservancy":
        "supporting_workers"
}

missing_columns = [
    col for col in COLUMN_MAPPING
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )

data = df.rename(columns=COLUMN_MAPPING).copy()

required_columns = [
    "conservancy_workers",
    "disposal_sites",
    "supporting_workers"
]

for column in required_columns:
    data[column] = pd.to_numeric(
        data[column],
        errors="coerce"
    )

if data[required_columns].isna().any().any():
    raise ValueError(
        "Dataset contains missing or invalid numeric values."
    )

if (data[required_columns] < 0).any().any():
    raise ValueError(
        "Workforce and disposal-site counts cannot be negative."
    )

if not np.all(
    data[required_columns].to_numpy()
    == np.floor(data[required_columns].to_numpy())
):
    raise ValueError(
        "Workforce and disposal-site counts must be integers."
    )

data[required_columns] = data[required_columns].astype(int)

if (data["disposal_sites"] == 0).any():
    raise ValueError(
        "At least one disposal site is required for allocation."
    )

data["total_workers"] = (
    data["conservancy_workers"]
    + data["supporting_workers"]
)

data["workers_per_site"] = (
    data["conservancy_workers"]
    / data["disposal_sites"]
)

data["supporting_workers_per_site"] = (
    data["supporting_workers"]
    / data["disposal_sites"]
)

data["total_workers_per_site"] = (
    data["total_workers"]
    / data["disposal_sites"]
)

data["conservancy_percentage"] = (
    data["conservancy_workers"]
    / data["total_workers"].replace(0, np.nan)
    * 100
).fillna(0)

data["supporting_percentage"] = (
    data["supporting_workers"]
    / data["total_workers"].replace(0, np.nan)
    * 100
).fillna(0)

print("\nProcessed Dataset:")
print(data.to_string(index=False))


# ============================================================
# 4. WORKFORCE STATISTICAL ANALYSIS
# ============================================================

total_conservancy = int(
    data["conservancy_workers"].sum()
)

total_supporting = int(
    data["supporting_workers"].sum()
)

total_sites = int(
    data["disposal_sites"].sum()
)

total_workforce = (
    total_conservancy + total_supporting
)

analysis = {
    "total_conservancy_workers": total_conservancy,
    "total_supporting_workers": total_supporting,
    "total_disposal_sites": total_sites,
    "total_workforce": total_workforce,

    "average_conservancy_workers_per_site": (
        total_conservancy / total_sites
    ),

    "average_supporting_workers_per_site": (
        total_supporting / total_sites
    ),

    "average_total_workers_per_site": (
        total_workforce / total_sites
    ),

    "conservancy_percentage": (
        100 * total_conservancy / total_workforce
        if total_workforce else 0
    ),

    "supporting_percentage": (
        100 * total_supporting / total_workforce
        if total_workforce else 0
    )
}

print("\n" + "=" * 65)
print("WORKFORCE ANALYSIS RESULTS")
print("=" * 65)

for key, value in analysis.items():
    print(f"{key}: {value:.4f}" if isinstance(
        value, float
    ) else f"{key}: {value}")


# ============================================================
# 5. MATHEMATICAL RESOURCE OPTIMIZATION MODEL
# ============================================================

class ConservancyResourceOptimizer:
    """
    Equal-distribution workforce allocation model.

    Assumptions:
    1. Every disposal site has equal workload.
    2. Every worker is available for allocation.
    3. Workers are assigned in whole numbers.
    4. No geographic or operational constraints are included.

    This is a mathematical optimization baseline,
    not a trained machine learning model.
    """

    def __init__(self):
        self.model_type = "Equal Distribution Optimizer"
        self.is_trained_ml_model = False
        self.allocation = None

    def allocate_workers(
        self,
        conservancy_workers,
        supporting_workers,
        disposal_sites
    ):

        if disposal_sites <= 0:
            raise ValueError(
                "Number of disposal sites must be positive."
            )

        if conservancy_workers < 0 or supporting_workers < 0:
            raise ValueError(
                "Worker counts cannot be negative."
            )

        def distribute(total, sites):
            base, remainder = divmod(total, sites)

            result = np.full(
                sites, base, dtype=np.int64
            )

            result[:remainder] += 1

            return result

        conservancy_allocation = distribute(
            conservancy_workers,
            disposal_sites
        )

        supporting_allocation = distribute(
            supporting_workers,
            disposal_sites
        )

        allocation_df = pd.DataFrame({
            "site_id": np.arange(1, disposal_sites + 1),
            "conservancy_workers": conservancy_allocation,
            "supporting_workers": supporting_allocation
        })

        allocation_df["total_workers"] = (
            allocation_df["conservancy_workers"]
            + allocation_df["supporting_workers"]
        )

        self.allocation = allocation_df

        return allocation_df

    def summary(self):

        if self.allocation is None:
            raise ValueError(
                "Run allocate_workers() first."
            )

        return {
            "model_type": self.model_type,
            "number_of_sites": int(len(self.allocation)),
            "total_allocated_workers": int(
                self.allocation["total_workers"].sum()
            ),
            "minimum_workers_per_site": int(
                self.allocation["total_workers"].min()
            ),
            "maximum_workers_per_site": int(
                self.allocation["total_workers"].max()
            ),
            "average_workers_per_site": float(
                self.allocation["total_workers"].mean()
            ),
            "allocation_standard_deviation": float(
                self.allocation["total_workers"].std(ddof=0)
            )
        }


optimizer = ConservancyResourceOptimizer()

allocation = optimizer.allocate_workers(
    conservancy_workers=total_conservancy,
    supporting_workers=total_supporting,
    disposal_sites=total_sites
)

optimization_summary = optimizer.summary()

assert (
    allocation["conservancy_workers"].sum()
    == total_conservancy
)

assert (
    allocation["supporting_workers"].sum()
    == total_supporting
)

print("\n" + "=" * 65)
print("RESOURCE ALLOCATION RESULTS")
print("=" * 65)

print("\nFirst 10 Sites:")
print(allocation.head(10).to_string(index=False))

print("\nOptimization Summary:")
for key, value in optimization_summary.items():
    print(f"{key}: {value}")


# ============================================================
# 6. SAVE H5 FILE
# ============================================================

def save_dataframe_h5(group, dataframe):

    for column in dataframe.columns:

        values = dataframe[column].to_numpy()

        if values.dtype.kind in ("O", "U", "S"):
            string_type = h5py.string_dtype(
                encoding="utf-8"
            )

            group.create_dataset(
                column,
                data=values.astype(str).tolist(),
                dtype=string_type
            )

        else:
            group.create_dataset(
                column,
                data=values
            )


with h5py.File(H5_PATH, "w") as h5:

    h5.attrs["project_name"] = (
        "AI-Based Conservancy Workforce Analysis"
    )

    h5.attrs["model_type"] = (
        "Mathematical Resource Allocation"
    )

    h5.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    h5.attrs["is_trained_ml_model"] = False

    dataset_group = h5.create_group(
        "processed_dataset"
    )

    save_dataframe_h5(
        dataset_group,
        data
    )

    allocation_group = h5.create_group(
        "resource_allocation"
    )

    save_dataframe_h5(
        allocation_group,
        allocation
    )

    statistics_group = h5.create_group(
        "workforce_statistics"
    )

    for key, value in analysis.items():
        statistics_group.attrs[key] = value

    optimization_group = h5.create_group(
        "optimization_summary"
    )

    for key, value in optimization_summary.items():
        optimization_group.attrs[key] = value

print("\nH5 File Saved:", H5_PATH)


# ============================================================
# 7. SAVE PKL MODEL
# ============================================================

model_package = {
    "optimizer": optimizer,
    "analysis": analysis,
    "optimization_summary": optimization_summary,
    "processed_data": data,
    "allocation": allocation,
    "model_type": "Mathematical Resource Allocation",
    "trained_machine_learning_model": False
}

with open(PKL_PATH, "wb") as file:
    pickle.dump(
        model_package,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print("\nPKL File Saved:", PKL_PATH)


# ============================================================
# 8. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based Conservancy Workforce Analysis, "
            "Waste Disposal Infrastructure Assessment, "
            "and Resource Optimization System"
        ),
        "version": "1.0",
        "dataset_year": "2017-18",
        "created_at": datetime.now().isoformat()
    },

    "dataset": {
        "path": str(CSV_PATH),
        "rows": int(df.shape[0]),
        "columns": int(df.shape[1]),
        "missing_values": int(df.isna().sum().sum())
    },

    "model": {
        "name": "ConservancyResourceOptimizer",
        "type": "Mathematical Optimization",
        "trained_ml_model": False,
        "allocation_strategy": "Equal Distribution",
        "objective": (
            "Balance workforce counts across disposal sites"
        )
    },

    "optimization": {
        "total_sites": total_sites,
        "conservancy_workers": total_conservancy,
        "supporting_workers": total_supporting,
        "total_workers": total_workforce,
        "equal_workload_assumption": True,
        "integer_allocation": True
    },

    "output_files": {
        "h5": H5_PATH.name,
        "pkl": PKL_PATH.name,
        "yaml": YAML_PATH.name,
        "json": JSON_PATH.name
    }
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML File Saved:", YAML_PATH)


# ============================================================
# 9. SAVE JSON METADATA
# ============================================================

metadata = {
    "project_name": (
        "AI-Based Conservancy Workforce Analysis, "
        "Waste Disposal Infrastructure Assessment, "
        "and Resource Optimization System"
    ),

    "dataset_year": "2017-18",

    "created_at": datetime.now().isoformat(),

    "dataset_information": {
        "filename": CSV_PATH.name,
        "records": int(df.shape[0]),
        "features": int(df.shape[1])
    },

    "workforce_analysis": analysis,

    "optimization_summary": optimization_summary,

    "model_information": {
        "model_name": "ConservancyResourceOptimizer",
        "model_type": "Equal Distribution Optimizer",
        "trained_machine_learning_model": False,
        "prediction_accuracy": None,
        "description": (
            "Mathematical workforce allocation baseline "
            "using aggregate workforce statistics."
        )
    },

    "allocation_results": allocation.to_dict(
        orient="records"
    ),

    "generated_files": {
        "h5": H5_PATH.name,
        "pkl": PKL_PATH.name,
        "yaml": YAML_PATH.name,
        "json": JSON_PATH.name
    }
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nJSON File Saved:", JSON_PATH)


# ============================================================
# 10. WORKFORCE VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

categories = [
    "Conservancy Workers",
    "Supporting Workers"
]

values = [
    total_conservancy,
    total_supporting
]

bars = plt.bar(categories, values)

plt.title("Conservancy Workforce Distribution")
plt.ylabel("Number of Workers")

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(int(height)),
        ha="center",
        va="bottom"
    )

plt.tight_layout()

graph_path = BASE_DIR / "workforce_distribution.png"

plt.savefig(
    graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. RESOURCE ALLOCATION VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    allocation["site_id"],
    allocation["conservancy_workers"],
    label="Conservancy Workers"
)

plt.plot(
    allocation["site_id"],
    allocation["supporting_workers"],
    label="Supporting Workers"
)

plt.title("Proposed Workforce Allocation Across Sites")
plt.xlabel("Disposal Site ID")
plt.ylabel("Allocated Workers")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

allocation_graph = BASE_DIR / "resource_allocation.png"

plt.savefig(
    allocation_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. VERIFY GENERATED FILES
# ============================================================

print("\n" + "=" * 65)
print("VERIFYING GENERATED FILES")
print("=" * 65)

# Verify H5
with h5py.File(H5_PATH, "r") as h5:
    print("\nH5 Groups:", list(h5.keys()))

# Verify PKL
with open(PKL_PATH, "rb") as file:
    loaded_package = pickle.load(file)

print(
    "PKL Model:",
    loaded_package["optimizer"].model_type
)

# Verify YAML
with open(YAML_PATH, "r", encoding="utf-8") as file:
    loaded_config = yaml.safe_load(file)

print(
    "YAML Project Version:",
    loaded_config["project"]["version"]
)

# Verify JSON
with open(JSON_PATH, "r", encoding="utf-8") as file:
    loaded_metadata = json.load(file)

print(
    "JSON Total Workforce:",
    loaded_metadata["workforce_analysis"]["total_workforce"]
)


# ============================================================
# 13. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("ALL FILES GENERATED SUCCESSFULLY")
print("=" * 65)

generated_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,
    graph_path,
    allocation_graph
]

for path in generated_files:

    if path.exists():
        print(
            f"[OK] {path.name} "
            f"({path.stat().st_size:,} bytes)"
        )

    else:
        print(f"[FAILED] {path.name}")

print("\nOutput Directory:")
print(BASE_DIR)

print("\nPROJECT EXECUTION COMPLETED")
```
