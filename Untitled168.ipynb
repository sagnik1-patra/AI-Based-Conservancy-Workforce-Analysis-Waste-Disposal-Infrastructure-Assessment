```python
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore", category=FutureWarning)

sns.set_theme(style="whitegrid")

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Conservancy Workforce Analysis, Waste Disposal Infrastructure Assessment, and Resource Optimization System"
)

CSV_PATH = BASE_DIR / (
    "number_of conservancy_workers_engaged_in_conservancy_"
    "garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"CSV file not found:\n{CSV_PATH}"
    )

print("=" * 70)
print("CONSERVANCY WORKFORCE ANALYSIS SYSTEM")
print("=" * 70)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.to_string(index=False))

print("\nDataset Shape:", df.shape)

# ============================================================
# 3. COLUMN VALIDATION
# ============================================================

COLUMN_MAPPING = {
    "Number of workers engaged in conservancy work":
        "conservancy_workers",

    "Number of garbage and sewage disposal dump site":
        "disposal_sites",

    "Number of persons engaged in other connected works related with conservancy":
        "supporting_workers"
}

missing = [
    col for col in COLUMN_MAPPING
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

data = df.rename(columns=COLUMN_MAPPING).copy()

numeric_columns = [
    "conservancy_workers",
    "disposal_sites",
    "supporting_workers"
]

for col in numeric_columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

if data[numeric_columns].isna().any().any():
    raise ValueError(
        "Required numeric columns contain missing values."
    )

if (data[numeric_columns] < 0).any().any():
    raise ValueError(
        "Worker and site counts cannot be negative."
    )

if not np.all(
    data[numeric_columns].to_numpy()
    == np.floor(data[numeric_columns].to_numpy())
):
    raise ValueError(
        "Worker and site counts must be integers."
    )

data[numeric_columns] = (
    data[numeric_columns].astype(int)
)

if (data["disposal_sites"] <= 0).any():
    raise ValueError(
        "Disposal site count must be positive."
    )

# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

data["total_workers"] = (
    data["conservancy_workers"]
    + data["supporting_workers"]
)

data["conservancy_workers_per_site"] = (
    data["conservancy_workers"]
    / data["disposal_sites"]
)

data["supporting_workers_per_site"] = (
    data["supporting_workers"]
    / data["disposal_sites"]
)

data["total_workers_per_site"] = (
    data["total_workers"]
    / data["disposal_sites"]
)

data["conservancy_percentage"] = (
    100
    * data["conservancy_workers"]
    / data["total_workers"].replace(0, np.nan)
).fillna(0)

data["supporting_percentage"] = (
    100
    * data["supporting_workers"]
    / data["total_workers"].replace(0, np.nan)
).fillna(0)

# ============================================================
# 5. AGGREGATE STATISTICS
# ============================================================

total_conservancy = int(
    data["conservancy_workers"].sum()
)

total_supporting = int(
    data["supporting_workers"].sum()
)

total_sites = int(
    data["disposal_sites"].sum()
)

total_workers = (
    total_conservancy + total_supporting
)

if total_sites <= 0:
    raise ValueError("No disposal sites available.")

print("\n" + "=" * 70)
print("WORKFORCE STATISTICS")
print("=" * 70)

print("Conservancy Workers:", total_conservancy)
print("Supporting Workers:", total_supporting)
print("Disposal Sites:", total_sites)
print("Total Workforce:", total_workers)

print(
    "Average Workers Per Site:",
    round(total_workers / total_sites, 2)
)

# ============================================================
# 6. RESOURCE ALLOCATION FUNCTION
# ============================================================

def equal_allocate(total, number_of_sites):
    """
    Distributes workers as evenly as possible.

    All worker counts are integers.
    The total number of workers is preserved.
    """

    base, remainder = divmod(
        int(total),
        int(number_of_sites)
    )

    allocation = np.full(
        number_of_sites,
        base,
        dtype=int
    )

    allocation[:remainder] += 1

    return allocation


conservancy_allocation = equal_allocate(
    total_conservancy,
    total_sites
)

supporting_allocation = equal_allocate(
    total_supporting,
    total_sites
)

allocation_df = pd.DataFrame({
    "site_id": np.arange(1, total_sites + 1),
    "conservancy_workers": conservancy_allocation,
    "supporting_workers": supporting_allocation
})

allocation_df["total_workers"] = (
    allocation_df["conservancy_workers"]
    + allocation_df["supporting_workers"]
)

assert (
    allocation_df["conservancy_workers"].sum()
    == total_conservancy
)

assert (
    allocation_df["supporting_workers"].sum()
    == total_supporting
)

allocation_df["allocation_method"] = (
    "Equal workload mathematical baseline"
)

allocation_df.to_csv(
    BASE_DIR / "resource_allocation.csv",
    index=False
)

print("\nResource Allocation:")
print(allocation_df.head(10).to_string(index=False))

# ============================================================
# 7. ACCURACY GRAPH
# ============================================================

# Accuracy, R2, MAE and RMSE require independent observed
# targets and sufficient train/test data.
#
# With one aggregate record, these metrics are unavailable.
# Do not create artificial test labels or fake accuracy.

fig, ax = plt.subplots(figsize=(10, 6))

ax.axis("off")

ax.text(
    0.5,
    0.78,
    "Machine Learning Evaluation",
    fontsize=19,
    fontweight="bold",
    ha="center",
    transform=ax.transAxes
)

ax.text(
    0.5,
    0.55,
    "Accuracy / R² / MAE / RMSE: N/A",
    fontsize=16,
    ha="center",
    transform=ax.transAxes
)

ax.text(
    0.5,
    0.35,
    "Only one aggregate observation is available.\n"
    "A valid train/test evaluation cannot be performed.",
    fontsize=12,
    ha="center",
    transform=ax.transAxes
)

ax.text(
    0.5,
    0.16,
    "Current method: Mathematical workforce allocation",
    fontsize=11,
    ha="center",
    transform=ax.transAxes
)

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAccuracy Graph Saved.")

# ============================================================
# 8. CORRELATION HEATMAP
# ============================================================

features = [
    "conservancy_workers",
    "supporting_workers",
    "disposal_sites",
    "total_workers",
    "conservancy_workers_per_site",
    "supporting_workers_per_site"
]

# Correlation requires variation across multiple observations.
# With one record, a Pearson correlation is undefined.

if len(data) >= 3:

    corr = data[features].corr()

    if corr.notna().any().any():

        plt.figure(figsize=(11, 8))

        sns.heatmap(
            corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f",
            vmin=-1,
            vmax=1,
            mask=corr.isna()
        )

        plt.title(
            "Workforce and Infrastructure Correlation Heatmap"
        )

        plt.tight_layout()

    else:
        fig, ax = plt.subplots(figsize=(10, 5))
        ax.axis("off")
        ax.text(
            0.5, 0.5,
            "Correlation unavailable:\n"
            "features have insufficient variation.",
            ha="center",
            va="center",
            fontsize=15
        )

else:

    # A descriptive matrix is valid for one observation.
    # Values are normalized to the largest value.
    descriptive_features = [
        "conservancy_workers",
        "supporting_workers",
        "disposal_sites",
        "total_workers"
    ]

    values = data[descriptive_features].sum()

    normalized = (
        values / max(values.max(), 1)
    )

    matrix = pd.DataFrame(
        [normalized.values],
        columns=[
            "Conservancy",
            "Supporting",
            "Disposal Sites",
            "Total Workers"
        ],
        index=["2017-18"]
    )

    plt.figure(figsize=(10, 4))

    sns.heatmap(
        matrix,
        annot=True,
        fmt=".3f",
        cmap="YlGnBu",
        vmin=0,
        vmax=1,
        linewidths=0.5
    )

    plt.title(
        "Normalized Workforce and Infrastructure Heatmap\n"
        "(Descriptive, Not Correlation)"
    )

    plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Heatmap Saved.")

# ============================================================
# 9. ALLOCATION COMPARISON
# ============================================================

# Compare two mathematical allocation strategies:
#
# A. Equal distribution
# B. Proportional distribution using illustrative
#    workload weights from 1 to 2.
#
# The weights are assumptions, not measured workloads.

def proportional_allocate(total, weights):

    weights = np.asarray(weights, dtype=float)

    if np.any(weights < 0) or weights.sum() <= 0:
        raise ValueError("Invalid workload weights.")

    raw = total * weights / weights.sum()

    allocation = np.floor(raw).astype(int)

    remainder = total - allocation.sum()

    fractional_parts = raw - allocation

    order = np.argsort(-fractional_parts)

    allocation[order[:remainder]] += 1

    return allocation


# Illustrative scenario, not observed site workloads.
workload_weights = np.linspace(
    1.0,
    2.0,
    total_sites
)

proportional_conservancy = proportional_allocate(
    total_conservancy,
    workload_weights
)

proportional_supporting = proportional_allocate(
    total_supporting,
    workload_weights
)

proportional_total = (
    proportional_conservancy
    + proportional_supporting
)

equal_total = (
    allocation_df["total_workers"].to_numpy()
)

# Compare deviation from hypothetical workload-based
# staffing targets, not real operational performance.

target_total = (
    total_workers
    * workload_weights
    / workload_weights.sum()
)

equal_error = np.mean(
    np.abs(equal_total - target_total)
)

proportional_error = np.mean(
    np.abs(proportional_total - target_total)
)

comparison_df = pd.DataFrame({
    "method": [
        "Equal Allocation",
        "Weighted Allocation"
    ],
    "mean_absolute_deviation_from_scenario_target": [
        equal_error,
        proportional_error
    ],
    "total_allocated_workers": [
        int(equal_total.sum()),
        int(proportional_total.sum())
    ]
})

comparison_df.to_csv(
    BASE_DIR / "allocation_comparison.csv",
    index=False
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    comparison_df["method"],
    comparison_df[
        "mean_absolute_deviation_from_scenario_target"
    ]
)

plt.ylabel("Mean Absolute Deviation (Workers)")

plt.title(
    "Allocation Strategy Comparison\n"
    "Illustrative Workload Scenario"
)

for bar in bars:

    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.3f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Comparison Graph Saved.")

# ============================================================
# 10. GENERATE RESULT CSV
# ============================================================

result_df = data.copy()

result_df["allocation_strategy"] = (
    "Equal Distribution"
)

result_df["allocated_conservancy_workers"] = (
    total_conservancy
)

result_df["allocated_supporting_workers"] = (
    total_supporting
)

result_df["allocated_total_workers"] = (
    total_workers
)

result_df["minimum_total_workers_per_site"] = int(
    allocation_df["total_workers"].min()
)

result_df["maximum_total_workers_per_site"] = int(
    allocation_df["total_workers"].max()
)

result_df["allocation_status"] = (
    "Mathematical baseline completed"
)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nRESULT CSV:")
print(result_df.to_string(index=False))

# ============================================================
# 11. RESULT GRAPH
# ============================================================

categories = [
    "Conservancy Workers",
    "Supporting Workers",
    "Total Workforce"
]

values = [
    total_conservancy,
    total_supporting,
    total_workers
]

plt.figure(figsize=(10, 6))

bars = plt.bar(categories, values)

plt.title(
    "Conservancy Workforce Analysis Results"
)

plt.xlabel("Workforce Category")
plt.ylabel("Number of Workers")

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Result Graph Saved.")

# ============================================================
# 12. SCENARIO-BASED WORKFORCE PROJECTION
# ============================================================

# These are illustrative future scenarios.
# They are not predictions learned from the dataset.
#
# Assumptions:
# - Disposal sites increase by 3% per year.
# - Workforce requirements increase by 4% per year.
# - The same proportional growth applies to both
#   conservancy and supporting workers.
#
# Change the assumptions to match actual planning data.

BASE_YEAR = 2018

FUTURE_YEARS = 10

SITE_GROWTH_RATE = 0.03

WORKFORCE_GROWTH_RATE = 0.04

future_years = np.arange(
    BASE_YEAR,
    BASE_YEAR + FUTURE_YEARS + 1
)

prediction_rows = []

for year in future_years:

    t = int(year - BASE_YEAR)

    projected_sites = int(
        round(
            total_sites
            * (1 + SITE_GROWTH_RATE) ** t
        )
    )

    projected_conservancy = int(
        round(
            total_conservancy
            * (1 + WORKFORCE_GROWTH_RATE) ** t
        )
    )

    projected_supporting = int(
        round(
            total_supporting
            * (1 + WORKFORCE_GROWTH_RATE) ** t
        )
    )

    projected_total = (
        projected_conservancy
        + projected_supporting
    )

    prediction_rows.append({
        "year": int(year),
        "projected_disposal_sites": projected_sites,
        "projected_conservancy_workers": (
            projected_conservancy
        ),
        "projected_supporting_workers": (
            projected_supporting
        ),
        "projected_total_workers": projected_total,
        "projected_workers_per_site": (
            projected_total / projected_sites
        ),
        "workforce_growth_assumption": (
            WORKFORCE_GROWTH_RATE
        ),
        "site_growth_assumption": (
            SITE_GROWTH_RATE
        ),
        "projection_type": (
            "Assumption-based scenario, not ML prediction"
        )
    })

prediction_df = pd.DataFrame(
    prediction_rows
)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nSCENARIO PROJECTIONS:")
print(prediction_df.to_string(index=False))

# ============================================================
# 13. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    prediction_df["year"],
    prediction_df["projected_conservancy_workers"],
    marker="o",
    linewidth=2,
    label="Conservancy Workers"
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_supporting_workers"],
    marker="s",
    linewidth=2,
    label="Supporting Workers"
)

plt.plot(
    prediction_df["year"],
    prediction_df["projected_total_workers"],
    marker="^",
    linewidth=2,
    label="Total Workforce"
)

plt.axvline(
    BASE_YEAR,
    linestyle="--",
    alpha=0.6,
    label="Baseline Year"
)

plt.title(
    "Scenario-Based Workforce Requirements\n"
    "Assumed Annual Workforce Growth: 4%"
)

plt.xlabel("Year")
plt.ylabel("Projected Number of Workers")

plt.xticks(future_years, rotation=45)

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Prediction Graph Saved.")

# ============================================================
# 14. OPTIONAL SITE-WISE ALLOCATION GRAPH
# ============================================================

plt.figure(figsize=(13, 6))

plt.plot(
    allocation_df["site_id"],
    allocation_df["conservancy_workers"],
    label="Conservancy Workers",
    linewidth=2
)

plt.plot(
    allocation_df["site_id"],
    allocation_df["supporting_workers"],
    label="Supporting Workers",
    linewidth=2
)

plt.plot(
    allocation_df["site_id"],
    allocation_df["total_workers"],
    label="Total Workers",
    linestyle="--",
    linewidth=2
)

plt.title(
    "Proposed Site-Wise Workforce Allocation"
)

plt.xlabel("Illustrative Disposal Site ID")
plt.ylabel("Allocated Workers")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "site_allocation_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 15. VERIFY OUTPUT FILES
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "resource_allocation.csv",
    "allocation_comparison.csv",
    "site_allocation_graph.png"
]

print("\n" + "=" * 70)
print("OUTPUT FILE VERIFICATION")
print("=" * 70)

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(
            f"[SUCCESS] {filename} "
            f"({path.stat().st_size:,} bytes)"
        )
    else:
        print(f"[FAILED] {filename}")

print("\nAll outputs saved to:")
print(BASE_DIR)

print("\nPROJECT EXECUTION COMPLETED SUCCESSFULLY!")
```
